```python
# 1. Write a Python program to find the "factorial of a number".

""" 
Enter a number : 5

 5 * 4 * 3 * 2 * 1  = 120
""" 
f = 1

number = int(input("Enter a number : "))  # 5

for i in range(1,number+1):  #  1 to 5 
    f *= i 

print(f"Factorial of {number} is {f} ")
```

```python
# 2. Write a Python program to check whether a "string is a palindrome".

s1 = input("Enter a string : ")

#print(s1[::-1])

rev = s1[::-1]
if rev == s1:
    print("is palindrome")
else:
    print("not a palindrome")

```

```python
# 3. Write a Python program to print the "Fibonacci series" up to `n` terms.

""" 
e.g.  Enter a number : 5

0  1  1  2  3

e.g.  Enter a number : 7

0  1  1  2  3  5  8
""" 

number = int(input("Enter a number : "))

first = 0 
second = 1 

print(f"{first}  {second}",end=" ")

for i in range(2,number):
    temp = first + second 
    print(temp,end=" ")
    first = second 
    second = temp

```

```python
# 4. Write a Python program to check whether a "number is prime".

""" 
prime number can only divide by 1 or itself 

5 :    1   5 

7 :    1   7 

6 :    1  2  3  6 
""" 

number = int(input("Enter a number : "))  #  7

if number <= 1:
    print("not a prime")
else:
    for i in range(2,number): # 2 3 4 5 6 
        if number % i == 0:
            print("not a prime number")
            break
    else:
        print("number is a prime number")
```
